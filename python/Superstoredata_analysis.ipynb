```python
import pandas as pd
#Load Datasets
df = pd.read_csv('SampleSuperstore.csv')
df.head()
```

```python
category_sales = df.groupby('Category')['Sales'].mean()
print(category_sales)
```

```python
#Which 5 cities have the highest TOTAL sales?
top_5_cities = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(5)
print(top_5_cities)
```
